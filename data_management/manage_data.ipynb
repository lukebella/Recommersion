# Managing data for training

```python
import pandas as pd 
import os
import librosa
```

Relative paths:

```python
base_path = "data/IEMOCAP_full_release/"

data_path = "dialog/EmoEvaluation/"
wav_path = "sentences/wav"

muse_path = "../data/c3_muse_trust/"
muse_wav_path = "audio_segments/"
muse_features = "label_segments/"
```

## MuSe

Retrieving the interested features from MuSe

```python
def muse_features_df(feature_path, feature_name) -> pd.DataFrame:
    feature_df = pd.DataFrame([], columns=["Name", "timestamp", feature_name])
    for file in os.listdir(feature_path):
        if file.endswith(".csv"):
            tmp_df = pd.read_csv(os.path.join(feature_path,file))
            if tmp_df["value"].isna().all():
                continue
            else:
                tmp_df = tmp_df[(tmp_df["value"] <= 1) & (tmp_df["value"]>= -1)]
                mean_df = tmp_df.groupby(tmp_df["segment_id"]).mean().reset_index()

                mean_df = mean_df.rename(columns={"value": feature_name, "segment_id": "Name"})
                mean_df["Name"] = mean_df["Name"].map(lambda x: file.rsplit(".", 1)[0]+"_"+str(x))
        
                feature_df = pd.concat([feature_df, mean_df],ignore_index=True)
    
    return feature_df

valence_path = os.path.join(muse_path,muse_features,"valence")
arousal_path = os.path.join(muse_path,muse_features,"arousal")

muse_val = muse_features_df(valence_path, "Valence")
muse_ar = muse_features_df(arousal_path, "Arousal")

muse_feat = pd.merge(muse_val, muse_ar, on=["Name", "timestamp"], how="inner")

print(muse_feat)
```

Retrieving and converting WAV files

```python
def retrieve_wav(path):
    wav_df = pd.DataFrame([], columns=["Name", "wav_file"])
    maj = 0
    for dir in os.listdir(path):
        for file in os.listdir(os.path.join(path, dir)):
            # Only proceed if the file is a .wav file
            if file.endswith(".wav"):
                file_path = os.path.join(path, dir, file)
                
                # Load the .wav file using librosa
                audio, _ = librosa.load(file_path, sr=16000)
                if len(audio) > 16000*8.75:
                    maj+=1
                # Add the audio array to the DataFrame
                wav_df = pd.concat([wav_df, pd.DataFrame([{"Name": file.replace(".wav",""), "wav_file" :audio}])])
    print("Files above 6 sec:", maj)
    return wav_df

muse_wav = retrieve_wav(os.path.join(muse_path, muse_wav_path))
muse_wav
```

Merging the two dataframes

```python
muse_full = pd.merge(muse_feat, muse_wav, on="Name", how="left")

# Drop rows that have NaN in 'wav_file' column (these are rows without matching .wav files)
muse_full = muse_full.dropna(subset=["wav_file"]).drop(columns="timestamp")

print("Final DataFrame with Valence, Arousal, and Wav:")
print(muse_full)
```

Normalizing dimensional values between 0 and 1

```python
muse_full["Valence"] =(muse_full["Valence"] - muse_full["Valence"].min()) / \
                      (muse_full["Valence"].max() - muse_full["Valence"].min())
muse_full["Arousal"] =(muse_full["Arousal"] - muse_full["Arousal"].min()) / \
                      (muse_full["Arousal"].max() - muse_full["Arousal"].min())

muse_full
```

```python
muse_full.to_pickle("../data/MuSe_sample")
```

## IEMOCAP

Managing and extracting the interested features

```python
import re

list_features = ["Start_Time", "End_Time", "Turn_Name", "Emotion", "Valence", "Arousal", "Dominance"]
df = pd.DataFrame([], columns=list_features)
for i in range(1,4):
    new_path = os.path.join(base_path,f'Session{i}/',data_path)
    for entry in os.listdir(new_path):
        if entry.endswith(".txt"):

            # Read the text file
            with open(new_path+entry, "r") as file:
                lines = file.readlines()

            # Regular expression to match lines starting with [START_TIME - END_TIME]
            pattern = r"\[(\d+\.\d+) - (\d+\.\d+)\]\s+(\S+)\s+(\S+)\s+\[(\d+\.\d+),\s*(\d+\.\d+),\s*(\d+\.\d+)\]"

            # List to hold the extracted data
            data = []

            # Loop through each line and match with the pattern
            for line in lines:
                match = re.match(pattern, line)
                if match:
                    start_time = float(match.group(1))
                    end_time = float(match.group(2))
                    turn_name = match.group(3)
                    emotion = match.group(4)
                    valence = float(match.group(5))
                    arousal = float(match.group(6))
                    dominance = float(match.group(7))
                    
                    # Append extracted data to the list
                    data.append([start_time, end_time, turn_name, emotion, valence, arousal, dominance])
            # Create a DataFrame from the extracted data
            temp_df = pd.DataFrame(data, columns=list_features)
            df = pd.concat([df, temp_df], ignore_index=True)

# Display the DataFrame
print(df)
```

Retrieving WAV files with Librosa

```python
wav_df = pd.DataFrame([], columns=["Turn_Name", "wav_file"])

for i in range(1, 6):
    session_path = os.path.join(base_path, f"Session{i}/", wav_path)
    tmp_path_list = os.listdir(session_path)
    
    # Iterate through directories within the session path
    for dir in tmp_path_list:

        dir_path = os.path.join(session_path, dir)
        
        if os.path.isdir(dir_path):
            for file in os.listdir(dir_path):
                if file.endswith(".wav"):
                    file_path = os.path.join(dir_path, file)
                    audio, sample_rate = librosa.load(file_path, sr=16000)
                    wav_df = pd.concat([wav_df, pd.DataFrame([{"Turn_Name": file.replace(".wav",""), "wav_file" :audio}])])

print(wav_df)
```

Merging features and files

```python
new_df = df.join(wav_df.set_index("Turn_Name"), "Turn_Name").drop(columns=["Start_Time", "End_Time", "Emotion"])
new_df
```

```python
def normalize(col, min, max):
    return (col - min)/(max - min)
```

```python
new_df["Arousal"] = normalize(new_df["Arousal"], new_df["Arousal"].min(), new_df["Arousal"].max())

new_df["Valence"] = normalize(new_df["Valence"], new_df["Valence"].min(), new_df["Valence"].max())

new_df["Dominance"] = normalize(new_df["Dominance"], new_df["Dominance"].min(), new_df["Dominance"].max())

new_df
```

```python
import IPython.display as ipd
import random

rand_int = random.randint(0, len(new_df))
print(new_df.iloc[rand_int])

ipd.Audio(data=new_df["wav_file"][rand_int], autoplay=True, rate=16000)
```

## MSP-PODCAST

```python
label_data = "../data/MSP/Labels/labels.txt"
speech_data = "../data/MSP/Audios/"
```

```python
import tarfile
import zipfile

data_path = "../data/MSP"

with zipfile.ZipFile('../data/MSP/Labels.zip', 'r') as zip_ref:
    zip_ref.extractall(data_path) 

print("Labels extraction complete.")

with tarfile.open('../data/MSP/Audios.tar.gz', 'r:gz') as tar:
    tar.extractall(data_path) 

print("Audios extraction complete.")
```

```python
import re

with open(label_data, "r") as f:
    features = [re.sub(r'\w+:', '',line.rstrip()).split('; ') for line in f if line.startswith("MSP-PODCAST_")]


print(features[0])
print(len(features))
```

```python
import pandas as pd

msp_features = pd.DataFrame(features, columns=["Name", "Emotion", "Arousal", "Valence", "Dominance"])
msp_features = msp_features.drop(columns=["Emotion", "Dominance"])
msp_features["Arousal"] = msp_features["Arousal"].astype(float)
msp_features["Valence"] = msp_features["Valence"].astype(float)

msp_features["Arousal"] = normalize(msp_features["Arousal"], msp_features["Arousal"].min(), msp_features["Arousal"].max())
msp_features["Valence"] = normalize(msp_features["Valence"], msp_features["Valence"].min(), msp_features["Valence"].max())

msp_features
```

```python
import os
import librosa

msp_wav = pd.DataFrame([], columns = ["Name", "wav_file"])
for file in os.listdir(speech_data):
    if file.endswith(".wav"):
        file_path = os.path.join(speech_data, file)
        audio, _ = librosa.load(file_path, sr=16000)
        msp_wav = pd.concat([msp_wav, pd.DataFrame([{"Name": file, "wav_file" :audio}])])

print(msp_wav)
```

```python
import IPython.display as ipd
import random

rand_int = random.randint(0, len(msp_wav))
print(msp_wav.iloc[rand_int])

ipd.Audio(data=msp_wav["wav_file"].iloc[rand_int], autoplay=True, rate=16000)
```

```python
msp = msp_features.join(msp_wav.set_index("Name"), on ="Name").dropna()
msp
```

```python
msp.to_pickle("../data/MSP_PODCAST")
```
