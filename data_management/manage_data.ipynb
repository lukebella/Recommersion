```python
import pandas as pd 
import os
import librosa
```

```python
base_path = "data/IEMOCAP_full_release/"

data_path = "dialog/EmoEvaluation/"
wav_path = "sentences/wav"

muse_path = "../data/c3_muse_trust/"
muse_wav_path = "audio_segments/"
muse_features = "label_segments/"
```

MuSe

```python
def muse_features_df(feature_path, feature_name) -> pd.DataFrame:
    feature_df = pd.DataFrame([], columns=["Name", "timestamp", feature_name])
    for file in os.listdir(feature_path):
        if file.endswith(".csv"):
            tmp_df = pd.read_csv(os.path.join(feature_path,file))#, columns=["timestamp", "value", "segment_id"])
            if tmp_df["value"].isna().all():
                continue
            else:
                tmp_df = tmp_df[(tmp_df["value"] <= 1) & (tmp_df["value"]>= -1)]
                mean_df = tmp_df.groupby(tmp_df["segment_id"]).mean().reset_index()

                mean_df = mean_df.rename(columns={"value": feature_name, "segment_id": "Name"})
                mean_df["Name"] = mean_df["Name"].map(lambda x: file.rsplit(".", 1)[0]+"_"+str(x))
        
                feature_df = pd.concat([feature_df, mean_df],ignore_index=True)
    
    return feature_df

valence_path = os.path.join(muse_path,muse_features,"valence")
arousal_path = os.path.join(muse_path,muse_features,"arousal")

muse_val = muse_features_df(valence_path, "Valence")
muse_ar = muse_features_df(arousal_path, "Arousal")

muse_feat = pd.merge(muse_val, muse_ar, on=["Name", "timestamp"], how="inner")

print(muse_feat)
```

```python
def retrieve_wav(path):
    wav_df = pd.DataFrame([], columns=["Name", "wav_file"])
    for dir in os.listdir(path):
        for file in os.listdir(os.path.join(path, dir)):
            # Only proceed if the file is a .wav file
            if file.endswith(".wav"):
                file_path = os.path.join(path, dir, file)
                
                # Load the .wav file using librosa
                audio, _ = librosa.load(file_path, sr=16000)
                # Add the audio array to the DataFrame
                wav_df = pd.concat([wav_df, pd.DataFrame([{"Name": file.replace(".wav",""), "wav_file" :audio}])])
    return wav_df

muse_wav = retrieve_wav(os.path.join(muse_path, muse_wav_path))
```

```python
muse_full = pd.merge(muse_feat, muse_wav, on="Name", how="left")

# Drop rows that have NaN in 'wav_file' column (these are rows without matching .wav files)
muse_full = muse_full.dropna(subset=["wav_file"]).drop(columns="timestamp")

print("Final DataFrame with Valence, Arousal, and Wav:")
print(muse_full)
```

```python
muse_full["Valence"] = muse_full["Valence"].apply(lambda x: (x+1)/2)
muse_full["Arousal"] = muse_full["Arousal"].apply(lambda x: (x+1)/2)

muse_full
```

```python
import re

list_features = ["Start_Time", "End_Time", "Turn_Name", "Emotion", "Valence", "Arousal", "Dominance"]
df = pd.DataFrame([], columns=list_features)
for i in range(1,6):
    new_path = os.path.join(base_path,f'Session{i}/',data_path)
    for entry in os.listdir(new_path):
        if entry.endswith(".txt"):

            # Read the text file
            with open(new_path+entry, "r") as file:
                lines = file.readlines()

            # Regular expression to match lines starting with [START_TIME - END_TIME]
            pattern = r"\[(\d+\.\d+) - (\d+\.\d+)\]\s+(\S+)\s+(\S+)\s+\[(\d+\.\d+),\s*(\d+\.\d+),\s*(\d+\.\d+)\]"

            # List to hold the extracted data
            data = []

            # Loop through each line and match with the pattern
            for line in lines:
                match = re.match(pattern, line)
                if match:
                    start_time = float(match.group(1))
                    end_time = float(match.group(2))
                    turn_name = match.group(3)
                    emotion = match.group(4)
                    valence = float(match.group(5))
                    arousal = float(match.group(6))
                    dominance = float(match.group(7))
                    
                    # Append extracted data to the list
                    data.append([start_time, end_time, turn_name, emotion, valence, arousal, dominance])
            # Create a DataFrame from the extracted data
            temp_df = pd.DataFrame(data, columns=list_features)
            df = pd.concat([df, temp_df], ignore_index=True)

# Display the DataFrame
print(df)
```
