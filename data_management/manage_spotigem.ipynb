```python
import pandas as pd
import os
```

```python
work_dir = os.getcwd().removesuffix("/data_management")
work_dir
```

```python
data_dir = work_dir+"/data/SpotigemDB"
data_dir
```

```python
artist = pd.read_csv(data_dir+"/artists_202404241006.csv", sep='\t', on_bad_lines="skip").dropna()
join = pd.read_csv(data_dir+"/join_artists_tracks_202404241007.csv", sep='\t').dropna()
tracks = pd.read_csv(data_dir+"/tracks_202404240955.csv", sep='\t', on_bad_lines="skip").dropna()

print(artist.columns)
print(join.columns)
print(tracks.columns)

```

```python
half_join = tracks.set_index("id").join(join.set_index("id_track"))\
                  .rename(columns={"uri":"uri_track", "popularity":"popularity_track"})
full_data = half_join.set_index("id_artist").join(artist.set_index("id"))
full_data.columns
```

```python
columns = ['uri_track', 'preview', 'title','energy','valence', 'name']
full_data = full_data[columns]
full_data
```

```python
print(full_data.iloc[0]["preview"])
```

```python
songs = pd.read_pickle("../data/Songs_path")
songs
```

```python
full_data.rename(columns={"uri_track":"musicId","arousal":"Arousal", "valence":"Valence",\
                    "name":"artist","preview":"mp3_path", "energy":"Arousal"} ,inplace=True)
full_data = full_data.dropna().drop_duplicates(subset=['title'], keep='last')
```

```python
full_data
```

```python
full = pd.concat([songs, full_data])
full
```

```python
full.to_pickle("../data/SpotiGEM_songs")
```
